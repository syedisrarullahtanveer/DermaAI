# DermaAI: Exploratory Data Analysis & Clinical Metadata Study

This notebook conducts in-depth demographic, statistical, and class distribution analysis on the **HAM10000** skin lesion dataset.

```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from src.config import load_config
from src.data.metadata_parser import ClinicalMetadataParser

config = load_config()
print(f"Project: {config.project.name} v{config.project.version}")
```

## 1. Metadata Ingestion and Validation
Inspect the schema, missing values, and column distributions.

```python
metadata_path = config.paths.metadata_csv
if os.path.exists(metadata_path):
    parser = ClinicalMetadataParser(metadata_path)
    df = parser.load_metadata()
    print(f"Loaded {len(df)} dermoscopy records.")
    display(df.head())
else:
    print(f"Metadata file not found at: {metadata_path}. Please place HAM10000_metadata.csv into dataset/metadata/")
```

## 2. Lesion Class Distribution & Class Imbalance Analysis

```python
if 'df' in locals():
    plt.figure(figsize=(10, 5))
    sns.countplot(data=df, x='dx', order=df['dx'].value_counts().index, palette='viridis')
    plt.title('HAM10000 Diagnostic Class Distribution')
    plt.xlabel('Diagnosis Code')
    plt.ylabel('Count')
    plt.grid(axis='y', alpha=0.3)
    plt.show()
```
